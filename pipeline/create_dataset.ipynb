```python
import gutenbergpy.textget
import pandas as pd
```

## Select the author with there book

```python
authors = {
    'H.P. Lovecraft': 74,            # Gothic horror (Obsessive detail, archaic vocabulary)
    'Jonathan Swift': 829,           # Gulliver's Travels (Satirical, long sentences)
    'Homer': 6130,                   # The Odyssey (Epic poetry, rhythmic and repetitive)
    'Virginia Woolf': 29220,         # Stream of consciousness (Modernist syntax)
    'Thomas Hardy': 110,             # Far from the Madding Crowd (Rural, descriptive)
    'Daniel Defoe': 521,             # Robinson Crusoe (Journalistic, plain style)
    'Agatha Christie': 1155          # The Secret Adversary (Mystery, heavy dialogue)
}
```

Download the books content, make data cleaning, chunking, and prepare for processing...

```python
dataset = []
```

```python
for name, book_id in authors.items():
    try:
        print(f"Downloading {name}...")
        raw_text = gutenbergpy.textget.get_text_by_id(book_id)
        clean_text = gutenbergpy.textget.strip_headers(raw_text).decode('utf-8')
        
        words = clean_text.split()
        for i in range(0, len(words), 500):
            chunk = " ".join(words[i : i + 500])
            if len(chunk.split()) > 100:
                dataset.append({'author': name, 'content': chunk})
    except Exception as e:
        print(f"Skipping {name} due to error: {e}")
```

```python
df = pd.DataFrame(dataset)
```
